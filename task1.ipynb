```python
from openai import OpenAI
```

```python
messages=[
        {
            "role": "user",
            "content": "what is 2 + 2? and give detailed answer"
        }
    ]
```

```python
client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)

response = client.chat.completions.create(
    model="gemma3:270m",
    messages=messages
)
    
print(response.choices[0].message.content)
```

### User prompts

```python
system_prompts = "You are a helpful assistant that give suitable replies to the user based on their input in a clear and concise manner."
```

```python
user_prompt = str(input("Enter your prompt: "))
```

```python
messages = [
    {"role": "system", "content": system_prompts},
    {"role": "user", "content": user_prompt}
]
```

```python
client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama"
)

response = client.chat.completions.create(
    model="llama3.2",
    messages=messages
)
    
print(response.choices[0].message.content)
```
