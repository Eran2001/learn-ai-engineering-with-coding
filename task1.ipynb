```python
import os

from dotenv import load_dotenv
from openai import (
    APIConnectionError,
    APITimeoutError,
    BadRequestError,
    InternalServerError,
    OpenAI,
    RateLimitError,
)
```

```python
load_dotenv(override=True)
```

```python
system_prompts = "You are a helpful assistant that translate English to Spanish."
```

```python
user_prompt = "How are you?"
```

```python
messages = [
    {"role": "system", "content": system_prompts},
    {"role": "user", "content": user_prompt},
]
```

```python
client = OpenAI(
    base_url="http://localhost:11434/v1",
    api_key="ollama",  # Ollama doesn't require a real key
)

try:
    response = client.chat.completions.create(model="llama3.2", messages=messages)
    choice = response.choices[0]
    if not choice.message.content:
        print("Empty reply, finish_reason:", choice.finish_reason)
    else:
        print(choice.message.content)
except (RateLimitError, APIConnectionError, APITimeoutError, InternalServerError) as e:
    print("Temporary error, still failed after retries:", e)
except BadRequestError as e:
    print("Problem with the request itself:", e)
```
